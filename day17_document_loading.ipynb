#读txt/md

```python
from langchain_core.documents import Document

def load_text(path:str)->Document:
    with open(path,encoding="utf-8")as f:
        text = f.read()
    return Document(page_content=text,metadata={"source":path})

doc = load_text("data/rag_notes_1.md")

print(doc.page_content[:100])
print(doc.metadata)
```

#读PDF

```python
from pypdf import PdfReader
from langchain_core.documents import Document

reader = PdfReader("data/test.pdf")
docs = []
for i,page in enumerate(reader.pages):
    text = page.extract_text() or ""

docs.append(Document(page_content=text,metadata={"source":"data/test.pdf","page":i}))

print(len(docs))  #页数
print(docs[0].metadata)
```

#批量加载多个文件

```python
import glob
from langchain_core.documents import Document

def load_text(path:str):
    with open(path,encoding="utf-8") as f:
        text = f.read()
    return Document(page_content=text,metadata={"source":path})

files = glob.glob("data/*.md")  #匹配data下所有.md文件
all_docs = [load_text(f) for f in files]

print("文件数:",len(files))
print("Document数:",len(all_docs))
for d in all_docs:
    print(d.metadata["source"])
```
