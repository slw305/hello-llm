#RecursiveCharacterTextSplitter基础用法

```python
import glob
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,  #每块目标长度
    chunk_overlap=50  #相邻块重叠50字符
)

def load_text(path:str):
    with open(path,encoding="utf-8") as f:
        text = f.read()
    return Document(page_content=text,metadata={"source":path})

files = glob.glob("data/*.md")  #匹配data下所有.md文件
all_docs = [load_text(f) for f in files]

chunks = splitter.split_documents(all_docs)

print(len(chunks))
print(chunks[0].page_content)
print(chunks[0].metadata)
```

```python
#不同chunk_size的直观感受
for s in [200,500,1000]:
    sp = RecursiveCharacterTextSplitter(chunk_size=s,chunk_overlap=50)
    cs = sp.split_documents(all_docs)
    print(f"chunk_size={s}:共{len(cs)}块,第1块{len(cs[0].page_content)}字符")
```
