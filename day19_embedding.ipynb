#用OpenAIEmbeddings接硅基流动BGE

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
import os

load_dotenv()

embeddings = OpenAIEmbeddings(
    model="BAAI/bge-m3",#硅基流动的免费中文embedding模型
    api_key=os.getenv("SILICONFLOW_API_KEY"),

    base_url="https://api.siliconflow.cn/v1"  #硅基流动的OpenAI兼容地址
)

vec = embeddings.embed_query("什么是RAG?")
print(len(vec))  #向量维度(bge-m3 = 1024)
print(vec[:5])  #前5个数(都是浮点数)
```

#验证"语义相近->向量相近"

```python
q = embeddings.embed_query("苹果很好吃")
a = embeddings.embed_query("苹果味道不错")
b = embeddings.embed_query("今天下大雨")
c = embeddings.embed_query("苹果很好吃")

from math import sqrt

def cosine(x,y):
    dot = sum(i * j for i,j in zip(x,y))  #点积
    nx = sqrt(sum(i*i for i in x))  # x的长度
    ny = sqrt(sum(i*i for i in y))  # y的长度
    return dot/(nx * ny)  #余弦相似度

print("苹果vs苹果：",round(cosine(q,a),4))  #应>0.9
print("苹果vs下雨：",round(cosine(q,b),4))  #应明显更低
print(round(cosine(q,c),4))
```
